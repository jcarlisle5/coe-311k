```python
import numpy as np
import matplotlib.pyplot as plt

k = 1
h_vals = [0.01,0.02,0.03,0.05,0.1,0.2,0.3,0.5]

x0 = 0
xf = 10

y0 = 1
v0 = 0

plt.figure()

for h in h_vals:

    print("\n----------------------")
    print("h =", h)

    # matrix (kinda just thrown together)
    A = np.array([[1, h],[-h*k, 1]])

    eig = np.linalg.eigvals(A)

    mags = np.abs(eig)

    print("eig:", eig)
    print("mag:", mags)

    stable = True
    for m in mags:
        if m >= 1:
            stable = False

    if stable == True:
        print("Solution is Stable")
    else:
        print("Solution may not be Entirely Stable ;)")

    # setup arrays
    x = np.arange(x0, xf+h, h)
    y = np.zeros(len(x))
    v = np.zeros(len(x))

    y[0] = y0
    v[0] = v0

    i = 1
    while i < len(x):
        y[i] = y[i-1] + h*v[i-1]
        v[i] = v[i-1] - h*k*y[i-1]
        i += 1

    plt.plot(x, y)

plt.title("Euler with diff h")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

```python
# The code builds the Euler update matrix for each step size, computes its eigenvalues, and checks their magnitudes to 
# assess stability. It then simulates the oscillator using Euler’s Method for each step size and plots the resulting 
# solution curves.
```
