```python
# Plot the data points and best fit line generated by the lest squared method

import numpy as np
import matplotlib.pyplot as plt

# Data points 
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
Y_fit = m * X + c
print(f"Slope: {m}, Intercept: {c}")
 
# plot the data points and best fit line
plt.scatter(X, Y, color = 'orange', label = 'Data points')
plt.plot(X, Y_fit, color = 'blue', label = f'Best fit: y = {m:.2f}x + {c:.2f}')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Data points and best fit line")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Part 2 (3 points) randomize the vectors, and rerun - how does it fit now?
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

Y = np.random.rand(5)
Y = Y * 6


A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
Y_fit = m * X + c
print(f"Slope: {m}, Intercept: {c}")
 
# plot the data points and best fit line
plt.scatter(X, Y, color = 'orange', label = 'Data points')
plt.plot(X, Y_fit, color = 'blue', label = f'Best fit: y = {m:.2f}x + {c:.2f}')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Data points and best fit line")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Part 3 (3 points) start with your own linear equation - 
# compute x (0 to 1, steps of .01) compute y using your own 
# linear equation +/- a random perturbation. Compute the Least 
# Squares and plot the best fit line, it *should* come 
# close to the original equation
```

```python
# Data points 
# Y_fit_orig = 2*x + 1
import random
X = np.arange(0, 1, 0.01)
Y = []

for i in range(len(X)):
    Y.append(2*X[i] + 1)
# print(Y)

for i in range(len(Y)):
    Y[i] = Y[i] + (np.random.uniform(-2, 2))
# print(Y)

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
Y_fit = m * X + c
print(f"Slope: {m}, Intercept: {c}")
```

```python
Y_fit = m * X + c
plt.scatter(X, Y, color = 'orange', label = 'Data points')
plt.plot(X, Y_fit, color = 'blue', label = f'Best fit: y = {m:.2f}x + {c:.2f}')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Data points and best fit line")
plt.legend()
plt.grid(True)
plt.show()
```
