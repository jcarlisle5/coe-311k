```python
## INSERT BASE 2 ALGORITHM HERE HAHAHHAHAHAHA


def base2(n):
    
    rem = 0 ## Remainder
    n = 2025 ## Initial number, Quotient
    remain_list = []
    running = True

    while running:
        rem = 1 if n % 2 == 1 else 0
        n = int(n/2)
        remain_list.append(rem)
        running = False if n == 0 else True
    print(remain_list[::-1])
    
base2(2025)
    
```
