```python
import numpy as np

# Define lambda(x)
def lam(x):
    return -2 * np.cos(x)

# Stability condition for Forward Euler:
# |1 + h*lambda| < 1 is worst case lambda magnitude

# Find worst case lambda over interval
x_vals = np.linspace(0, 2*np.pi, 1000) # create array
lambda_vals = lam(x_vals) # multiply them

lambda_min = np.min(lambda_vals) # return min
lambda_max = np.max(lambda_vals) # return max

print("Lambda range:", lambda_min, "to", lambda_max) # print them

# Worst-case magnitude (most negative value controls stability)
lambda_worst = min(lambda_vals)  # most negative

# Stability condition: -2 < h*lambda < 0
# h < 2 / |lambda_worst| DOES THIS
h_max = 2 / abs(lambda_worst)

print("Maximum stable step size h:", h_max)
```
