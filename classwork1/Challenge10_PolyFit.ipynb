```python
# Part A

# Polyfit function

import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)

# Step 1: Generate some sample data
# Creating data points that roughly follow a quadratic curve

x = np.linspace(-5, 5, 50) # 50 points between -5 and 5
y = 2 * x**3 + 3 * x + np.random.randn(50) * 10 # y = 2x^2 + 3x + some noise

# Step 2: Polynomial fitting using NumPy's polyfit
# Fit a 2nd degree polynomial to the data
degree = 3 # Degree of the polynomial
coefficients = np.polyfit(x, y, degree)
# Print the polynomial coefficients
print("Polynomial Coefficients: ", coefficients)
# Step 3: Use the coefficients to predict y-values (polynomial function)
# Generate polynomial function based on the coefficients
polynomial = np.poly1d(coefficients)

# Generate predicted y-values based on the fitted polynomial
y_pred = polynomial(x)
# Step 4: Plotting the results
plt.scatter(x, y, color='blue', label='Data points') # Original data points
plt.plot(x, y_pred, color='red', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()

# Mean Squared Error (MSE)
mse1 = np.mean((y - y_pred)**2)
print(f'Mean Squared Error: {mse1:.4f}')
# R-squared value
ss_total = np.sum((y - np.mean(y))**2)
ss_res = np.sum((y - y_pred)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared: {r_squared:.4f}')
```

```python
# Part A Least Square

# Data points 
X = np.linspace(-5, 5, 50) # 50 points between -5 and 5
Y = 2 * x**3 + 3 * x + np.random.randn(50) * 10 # y = 2x^2 + 3x + some noise

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
Y_fit = m * X + c
print(f"Slope: {m}, Intercept: {c}")
 
# plot the data points and best fit line
plt.scatter(X, Y, color = 'orange', label = 'Data points')
plt.plot(X, Y_fit, color = 'blue', label = f'Best fit: y = {m:.2f}x + {c:.2f}')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Data points and best fit line")
plt.legend()
plt.grid(True)
plt.show()

mse = np.mean((Y - Y_fit)**2)

# R^2 calculation
ss_res = np.sum((Y - Y_fit)**2)          # residual sum of squares
ss_tot = np.sum((Y - np.mean(Y))**2)     # total sum of squares
r2 = 1 - (ss_res / ss_tot)

print(f"Mean Squared Error: {mse}")
print(f"R^2: {r2}")
```

```python
if mse < mse1:
    print("Least Squares Approximation had a lower MSE, so it's more accurate. ")
else:
    print("Polyfit had a lower MSE, so it's more accurate.")
    print()
if r2 > r_squared:
    print("Least Squares Approximation had a higher r^2, so it's more accurate. ")
else:
    print("Polyfit had a higher r^2, so it's more accurate.")
    print()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data points
x = np.linspace(-5, 5, 50)
y = 2 * x**3 + 3 * x + np.random.randn(50) * 10  # noisy cubic

# Degree of the polynomial
degree = 3

# Create Vandermonde matrix (lowest to highest power)
V = np.vander(x, N=degree+1, increasing=True)

# Solve for coefficients using least squares
coefficients = np.linalg.lstsq(V, y, rcond=None)[0]

# Reverse coefficients for standard polynomial order (highest to lowest)
coefficients_standard = coefficients[::-1]
print("Polynomial Coefficients (highest degree first):", coefficients_standard)

# Compute fitted values
y_pred = V @ coefficients

# Calculate Mean Squared Error (MSE)
mse = np.mean((y - y_pred) ** 2)
print("Mean Squared Error (MSE):", mse)

# Calculate R-squared (coefficient of determination)
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r_squared = 1 - (ss_res / ss_tot)
print("R-squared (R^2):", r_squared)

# Plot
plt.scatter(x, y, label="Data points")
plt.plot(x, y_pred, color="red", label=f"Degree {degree} fit")
plt.legend()
plt.show()
```
