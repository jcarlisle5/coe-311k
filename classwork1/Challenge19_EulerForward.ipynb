```python
import numpy as np
import matplotlib.pyplot as plt

h = 0.0005

t0 = 0
tf = 2

t = np.arange(t0, tf+h, h)

y = np.zeros(len(t))
y[0] = 0  # initial condition

# Euler Forward
i = 1
while i < len(t):
    y[i] = y[i-1] + h * (-1000*y[i-1] + 3000 - 2000*np.exp(-t[i-1]))
    i += 1

# analytical solution
y_true = 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

# plot
plt.plot(t, y, label="Euler")
plt.plot(t, y_true, '--', label="Analytical")

plt.title("Euler vs Analytical (stiff ODE)")
plt.xlabel("t")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```
