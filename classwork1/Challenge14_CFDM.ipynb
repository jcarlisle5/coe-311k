```python
# Approximating Derivative of f(x)=sin⁡(x) with 
# Forward Difference, Central Difference, Backwards Difference

import math
import numpy as np
```

```python
# Step size, setup initial values

a = 0
b = (2 * np.pi)
f = np.sin
N = 50

x = np.linspace(a, b, N)
h = x[1] - x[0]

# Central Difference Approximation for the derivative
# df_dx = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approx.
# df_dx = (f(x + h) − f(x))/h

# Backwards Difference Approx.
#df_dx = (f(x) − f(x − h))/h

```

```python
df_forward = np.zeros(N)
df_forward[:-1] = (f(x[1:]) - f(x[:-1])) / h  # all except last point
df_forward[-1] = (f(x[-1]) - f(x[-2])) / h    # use backward at last point

# Backward
df_backward = np.zeros(N)
df_backward[1:] = (f(x[1:]) - f(x[:-1])) / h  # all except first point
df_backward[0] = (f(x[1]) - f(x[0])) / h      # use forward at first point

# Central
df_central = np.zeros(N)
df_central[1:-1] = (f(x[2:]) - f(x[:-2])) / (2*h)  # middle points
df_central[0] = (f(x[1]) - f(x[0])) / h           # forward for first point
df_central[-1] = (f(x[-1]) - f(x[-2])) / h        # backward for last point

# actual derivative
df_exact = np.cos(x)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, df_exact, 'k-', label='Exact cos(x)')
plt.plot(x, df_forward, 'r--', label='Forward Difference')
plt.plot(x, df_backward, 'b--', label='Backward Difference')
plt.plot(x, df_central, 'g--', label='Central Difference')
plt.legend()
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Derivative Approximations of sin(x)")
plt.show()
```
