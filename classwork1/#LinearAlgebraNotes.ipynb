```python
import numpy as np

## how to create a vector

v1 = np.array([1, 3, 5, 7, 9, 11])
v2 = np.array([2, 4, 6, 8, 10, 12])

print(f"Vector 1: {v1}")
print(f"Vector 2: {v2}")

## how to add vectors

v_add = v1 + v2
v_subtract = v1 - v2

print(f"v1+v2 = {v_add}")
print(f"v1-v2 = {v_subtract}")
```

```python
## how to scale vectors

scalar = 2
v_scaled = scalar * v1
print(f"v1 * scalar = {v_scaled}")
```

```python
## how to dot vectors

dot_product = np.dot(v1, v2)
print(f"dot product v1, v2 = {dot_product}")
```

```python
## Norm of a vector

magnitude_v1 = np.linalg.norm(v1)
print(f"v1 norm is = {magnitude_v1:.4f}")
```

```python
## multi dimensional array

A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)
print()
B = np.array([[7, 8, 9], [10, 11, 12], [13, 14, 15]])
print(B)
```

```python
## adding matricies

C = np.array([[1, 2, 3], [4, 5, 6,]])
matrix_add = A + C
matrix_subtract = A - C
print(f"matrixA + matrixC = {matrix_add}")
print(f"matrixA - matrixC = {matrix_subtract}")
```

```python
## multiplying matricies

matrix_mul = np.dot(A, B)
print(f"matrixA * matrixB = {matrix_mul}")
```

```python
## Transposing

transpose_A = A.T
print(f"Transpose_A = {transpose_A}")
```

```python
## solving matricies
A = np.array([[2, 3], [4, -1]])
b = np.array([5, 3])
print(A)
print(B)
x = np.linalg.solve(A, b)
print(x)
```

```python
## eiganvalues and eiganvectors

A = np.array([[2, 1], [1, 2]])
e_value, e_vector = np.linalg.eig(A)
print(f"eiganvalue and eiganvector = {e_value}, {e_vector}")
```

```python
A = np.array([[4, 7], [2, 6]])
A_inv = np.linalg.inv(A)

print(A)
print(f"A_inverse = {A_inv}")
```

```python
## identity matrix

identity = np.dot(A, A_inv)
print(identity)
```
