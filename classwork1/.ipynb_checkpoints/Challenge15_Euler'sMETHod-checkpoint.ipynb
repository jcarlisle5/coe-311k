```python
#loop(range) your step size to go from .01 to 1, steps of .05
# Plot each step-size run, keep an eye on how the error is related to the 
# step-size, print the error(current run vs first run) for each point, 
# and the total accumulated error for each run.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

# Initial condition
x_start, x_end = 0, 2
y_values_first_run = None  # to compare errors

# Loop over step sizes
for h in np.arange(0.01, 1.01, 0.05):  # step sizes from 0.01 to 1, step 0.05
    num_steps = int((x_end - x_start) / h)
    
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    
    y_values[0] = 1  # y(0) = 1
    
    # Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    
    # Store first run for error comparison
    if y_values_first_run is None:
        y_values_first_run = y_values.copy()
    
    # Compute error vs first run
    min_len = min(len(y_values_first_run), len(y_values))
    error = np.abs(y_values[:min_len] - y_values_first_run[:min_len])
    
    print(f"Step size {h:.2f} -> Total accumulated error: {np.sum(error):.5f}")
    
    # Plot each run
    plt.plot(x_values, y_values, label=f"h={h:.2f}")

plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler's Method Approximation of dy/dx = -2y for Various Step Sizes")
plt.legend()
plt.show()
```
